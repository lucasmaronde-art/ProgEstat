```python
distribuicoes = {
    "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
    "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
    "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
}
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
distribuicoes["Normal"]["dados"] = np.random.normal(loc=0, scale=1, size=1000)
```

```python
distribuicoes["Exponencial"]["dados"] = np.random.exponential(scale =2, size =1000)
```

```python
distribuicoes["Uniforme"]["dados"]= np.random.uniform(low=-1, high=1, size = 1000)
```

```python
# Normal
media = np.mean(distribuicoes["Normal"]["dados"])
median = np.median(distribuicoes["Normal"]["dados"])
variance = np.var(distribuicoes["Normal"]["dados"])
std = np.std(distribuicoes["Normal"]["dados"])
distribuicoes["Normal"]["estatisticas"] = {"media": media, "median": median, "variance": variance,"std": std}
```

```python
#Exponential
media = np.mean(distribuicoes["Exponencial"]["dados"])
median = np.median(distribuicoes["Exponencial"]["dados"])
variance = np.var(distribuicoes["Exponencial"]["dados"])
std = np.std(distribuicoes["Exponencial"]["dados"])
distribuicoes["Exponencial"]["estatisticas"] = {"media": media, "median": median, "variance": variance,"std": std}
```

```python
#Uniform
media = np.mean(distribuicoes["Uniforme"]["dados"])
median = np.median(distribuicoes["Uniforme"]["dados"])
variance = np.var(distribuicoes["Uniforme"]["dados"])
std = np.std(distribuicoes["Uniforme"]["dados"])
distribuicoes["Uniforme"]["estatisticas"] = {"media": media, "median": median, "variance": variance,"std": std}
```

```python
#Passo 7

distribuicoes["Poisson"] = {"params": {"scale": 3}, "dados": None, "estatisticas": None}
distribuicoes["Poisson"]["dados"] = np.random.poisson(lam = 3, size = 1000)
media = np.mean(distribuicoes["Poisson"]["dados"])
median = np.median(distribuicoes["Poisson"]["dados"])
variance = np.var(distribuicoes["Poisson"]["dados"])
std = np.std(distribuicoes["Poisson"]["dados"])
distribuicoes["Poisson"]["estatisticas"] = {"media": media, "median": median, "variance": variance,"std": std}
df["Poisson"] = distribuicoes["Poisson"]["estatisticas"]
```

A média da Normal chegou a -0,041679 e sua esperança é 0. Logo, está próximo de sua esperança. A da Uniforme apresenta estar próxima também, uma vez que a esperança é 0 e a média encontrada foi 0,037. Na exponencial também está próxima, já que sua esperança é 2 e a média apresentada foi 2,02



```python
#Passo 8

for dist in distribuicoes:
  distribuicoes[dist]["estatisticas"]["min"] = np.min(distribuicoes[dist]["dados"])
  distribuicoes[dist]["estatisticas"]["max"] = np.max(distribuicoes[dist]["dados"])
  distribuicoes[dist]["estatisticas"]["quantis"] = np.percentile(distribuicoes[dist]["dados"], [25, 50, 75])
```

```python
dados_plot = [distribuicoes[dist]["dados"] for dist in distribuicoes]
nomes_plot = [dist for dist in distribuicoes]
fig, ax = plt.subplots(figsize=(10, 6))
ax.boxplot(
    x=dados_plot,
    labels=nomes_plot,
    vert=True,
    patch_artist=True,
    notch=False,
    showmeans=True,
    showfliers=True,
    widths=0.6
)
```

```python
#Passo 10

media_teorica_normal = distribuicoes["Normal"]["params"]["loc"]
media_teorica_exponencial = distribuicoes["Exponencial"]["params"]["scale"]
media_teorica_uniforme = (distribuicoes["Uniforme"]["params"]["high"]+distribuicoes["Uniforme"]["params"]["low"])/2
media_teorica_poisson = distribuicoes["Poisson"]["params"]["scale"]

nomes = []
medias_teoricas = [media_teorica_normal, media_teorica_exponencial, media_teorica_uniforme, media_teorica_poisson]
medias_amostra = []

for dist in distribuicoes:
  nomes.append(dist)
  media_amo = np.mean(distribuicoes[dist]["dados"])
  medias_amostra.append(media_amo)

x = np.arange(len(nomes))  # Posições das distribuições no eixo X
largura = 0.35             # Largura de cada barra

fig, ax = plt.subplots(figsize=(8, 6))

# Criando as barras lado a lado (deslocando metade da largura para esquerda/direita)
barras_teoricas = ax.bar(x - largura/2, medias_teoricas, largura, label='Média Teórica', color='cornflowerblue')
barras_amostra = ax.bar(x + largura/2, medias_amostra, largura, label='Média da Amostra', color='lightcoral')

# Personalizando o gráfico
ax.set_ylabel('Valores Médios', fontsize=12)
ax.set_title('Comparação: Média Teórica vs. Média da Amostra', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(nomes, fontsize=11)
ax.legend()

# Adicionando uma grade no fundo para facilitar a leitura
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
